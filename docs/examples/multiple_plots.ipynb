# Multiple plots in one notebook

Each `display_model` call (`ball_and_stick`, `stick`, `van_der_waals`, `sas`, `ses`) gets its own DOM-scoped event channel, so several plots can coexist in one notebook and stay independently interactive.

Call `Bonito.Page()` once near the top so the (large) JavaScript bundle is shipped only once for the whole notebook instead of being re-embedded into every cell.

```julia
import Pkg
Pkg.activate("../..")
using BiochemicalAlgorithms, BiochemicalVisualization, Bonito
Page()
```
